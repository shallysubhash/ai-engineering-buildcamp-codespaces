```python
from pathlib import Path
from urllib.request import urlopen
from urllib.parse import urlparse
import csv

def download_file(url, destination):
    with urlopen(url, timeout=120) as response:
        data = response.read()

    destination.write_bytes(data)


csv_url = (
    "https://raw.githubusercontent.com/"
    "alexeygrigorev/ai-engineering-buildcamp-code/"
    "main/01-foundation/homework/books.csv"
)

csv_path = Path("books.csv")
books_dir = Path("books")
books_dir.mkdir(exist_ok=True)

download_file(csv_url, csv_path)

print("CSV downloaded.")
print("Books will be saved in:", books_dir.resolve())
with csv_path.open(encoding="utf-8-sig", newline="") as file:
    books = list(csv.DictReader(file))

for book in books:
    print(book["title"], "→", book["pdf_url"])

print("\nNumber of books:", len(books))
failed_downloads = []

for book in books:
    title = book["title"]
    pdf_url = book["pdf_url"]

    # Extract the filename from the URL
    filename = Path(urlparse(pdf_url).path).name
    destination = books_dir / filename

    try:
        print(f"Downloading: {title}")
        download_file(pdf_url, destination)
        print(f"Saved: {destination}\n")

    except Exception as error:
        failed_downloads.append(title)
        print(f"Failed: {title} — {error}\n")

if failed_downloads:
    print("Downloads that failed:", failed_downloads)
else:
    print(f"All {len(books)} PDFs downloaded successfully!")

```
