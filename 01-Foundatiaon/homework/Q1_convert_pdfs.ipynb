```python
from pathlib import Path
from markitdown import MarkItDown

books_dir = Path("books")
text_dir = Path("books_text")
text_dir.mkdir(exist_ok=True)

pdf_files = sorted(books_dir.glob("*.pdf"))

print("Working directory:", Path.cwd())
print("PDFs found:", len(pdf_files))

if not pdf_files:
    raise FileNotFoundError(
        "No PDFs found. Check that this notebook runs "
        "from the folder containing books/."
    )
converter = MarkItDown()
failed = []

for pdf_path in pdf_files:
    markdown_path = text_dir / f"{pdf_path.stem}.md"

    try:
        print(f"Converting: {pdf_path.name}")

        result = converter.convert(str(pdf_path))
        markdown_path.write_text(
            result.text_content,
            encoding="utf-8",
        )

        print(f"Saved: {markdown_path}\n")

    except Exception as error:
        failed.append(pdf_path.name)
        print(f"Failed: {pdf_path.name} — {error}\n")

print("Failed conversions:", failed)

think_python_path = text_dir / "thinkpython2.md"

content = think_python_path.read_text(encoding="utf-8")
line_count = len(content.splitlines())

print("Think Python, 2nd edition:", line_count, "lines")
```
