```python
from pathlib import Path
from gitsource import chunk_documents

text_dir = Path("books_text")
markdown_files = sorted(text_dir.glob("*.md"))

if not markdown_files:
    raise FileNotFoundError(
        "No Markdown files found in books_text/. "
        "Check your notebook's working directory."
    )

documents = []

for file_path in markdown_files:
    text = file_path.read_text(encoding="utf-8")

    non_empty_lines = [
        line
        for line in text.splitlines()
        if line.strip()
    ]

    documents.append({
        "source": file_path.name,
        "content": non_empty_lines,
    })

    print(f"{file_path.name}: {len(non_empty_lines)} non-empty lines")

    chunks = chunk_documents(
    documents,
    size=100,
    step=50,
)

print("Total chunks across all books:", len(chunks))

think_python_chunks = [
    chunk
    for chunk in chunks
    if chunk["source"] == "thinkpython2.md"
]

print(
    "Think Python, 2nd edition:",
    len(think_python_chunks),
    "chunks",
)

first_chunk = think_python_chunks[0]

print("Source:", first_chunk["source"])
print("Starting position:", first_chunk["start"])
print("Lines in chunk:", len(first_chunk["content"]))
print("\nPreview:\n")
print("\n".join(first_chunk["content"][:10]))
```
